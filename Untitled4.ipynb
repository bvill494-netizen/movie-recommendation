### Sample Live Retrieval Function from TMDB API

Below is an example of how you can query TMDB's popular movies endpoint using your active API key.

```python
def fetch_popular_movies(api_key):
    url = f"https://api.themoviedb.org/3/movie/popular?api_key={api_key}&language=en-US&page=1"
    response = requests.get(url)
    if response.status_code == 200:
        results = response.json().get('results', [])
        # Create a simplified list of movies similar to our local dataset schema
        live_movies = []
        for movie in results:
            live_movies.append({
                "title": movie.get('title'),
                "plot": movie.get('overview'),
                "rating": movie.get('vote_average'),
                "vote_count": movie.get('vote_count'),
                "release_year": int(movie.get('release_date', '0000')[:4]) if movie.get('release_date') else 0
            })
        return pd.DataFrame(live_movies)
    else:
        print(f"Failed to connect to TMDB: {response.status_code}")
        return None

# Uncomment the lines below to test with your actual TMDB Key:
# df_live = fetch_popular_movies(TMDB_API_KEY)
# if df_live is not None:
#     display(df_live.head())
```

### Content-Based & Sentiment Recommendation Engine
We will build a hybrid matching function that combines plot text TF-IDF similarity with sentiment analysis matches.

```python
# Initialize TF-IDF Vectorizer
tfidf = TfidfVectorizer(stop_words='english')
tfidf_matrix = tfidf.fit_transform(df['plot'])

def recommend_movies(query, min_rating=0.0, max_runtime=200, selected_genre="All", sentiment_preference="Any"):
    # 1. Start with initial data copying
    filtered_df = df.copy()

    # 2. Apply metadata filters
    filtered_df = filtered_df[filtered_df['rating'] >= min_rating]
    filtered_df = filtered_df[filtered_df['runtime'] <= max_runtime]

    if selected_genre != "All":
        filtered_df = filtered_df[filtered_df['genre'].str.contains(selected_genre, case=False)]

    if filtered_df.empty:
        return "No movies match the physical/genre criteria."

    # 3. Calculate Content Similarity with User Query using NLP
    query_vector = tfidf.transform([query])
    filtered_indices = filtered_df.index
    similarities = cosine_similarity(query_vector, tfidf_matrix[filtered_indices]).flatten()

    # Add content similarity to the filtered dataframe view
    filtered_df['similarity_score'] = similarities

    # 4. Filter or adjust by Sentiment Preference
    # Sentiment scores range from -1 (very negative) to 1 (very positive)
    if sentiment_preference == "Positive/Uplifting":
        filtered_df = filtered_df[filtered_df['sentiment_score'] > 0.1]
    elif sentiment_preference == "Dark/Intense":
        filtered_df = filtered_df[filtered_df['sentiment_score'] <= 0.1]

    if filtered_df.empty:
        return "No movies matching that sentiment category met your metadata criteria."

    # Sort by the NLP similarity score
    results = filtered_df.sort_values(by='similarity_score', ascending=False).head(3)

    # Format output for presentation
    output_lines = []
    for idx, row in results.iterrows():
        sentiment_desc = "Positive/Uplifting" if row['sentiment_score'] > 0.1 else "Dark/Intense"
        output_lines.append(f"### **{row['title']}** ({row['release_year']}) - {row['rating']}⭐")
        output_lines.append(f"* **Genre:** {row['genre']} | **Length:** {row['runtime']} min | **Cert:** {row['certification']}")
        output_lines.append(f"* **Plot Summary:** {row['plot']}")
        output_lines.append(f"* **NLP Evaluation:** Similarity Match: {row['similarity_score']:.2f} | Tone: {sentiment_desc} ({row['sentiment_score']:.2f})")
        output_lines.append("---")

    return "\n".join(output_lines)
```
